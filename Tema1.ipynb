```python
#ex3 functie de reducere la intervalul (-pi/2,pi/2)

def reducere(x):
    x=(x+pi/2)%pi-pi/2
 
    if(abs(abs(x)-pi/2)<10**(-12)):
        return float('nan')
        
    return x
    
```

```python
#ex3 tan(x) folosind fractii continue

def my_tan_fractii_continue(x, epsilon):

    if(x>pi/2 or x<-pi/2):
        x=reducere(x)

    if(isnan(x)):
        return float('nan')

    mic=10**(-12)
    f=mic
    C=f
    D=0
    k=1
    x3=x**2
    
    while True:
       
        if(k==1):
            aj=x
        else:
            aj=-x3

        bj=2*k-1
        
        D=bj+aj*D
        if(D==0):
            D=mic

        C=bj+aj/C
        if(C==0):
            C=mic

        D=1/D
        delta=C*D
        
        f=delta*f
        k=k+1
        
        if(abs(delta-1)<epsilon):
            return f
```

```python
#ex3 tan(x) folosind polinoame 

def my_tan_polinom(x):
    c1 = 0.33333333333333333   
    c2=0.133333333333333333  
    c3=0.053968253968254    
    c4=0.0218694885361552   
    
    x2=x*x
    tan=x*(1.0+x2*(c1+x2*(c2+x2*(c3+x2*c4))))
    
    return tan


def preprocesare(x):

    if(x>pi/2 or x<(-pi/2)):
      x=reducere(x)

    if(isnan(x)):
       return float('nan')
    
    sign=1
    if(x<0):
        sign=-1
        x=-x
    
    reduc_interval=False
    
    # tan(x) = 1/tan(pi/2-x)
    if (x>=pi/4):
        x=pi/2-x
        reduc_interval=True
    
    result=my_tan_polinom(x)
    
    if(reduc_interval):
        result=1/result

    return sign*result
```

```python
#main ex3

n=10000
random.seed(42)
values=[random.uniform(-pi/2+10**(-15), pi/2-10**(-15)) for _ in range(n)]

t1_start=time.perf_counter()
results_fractii=[my_tan_fractii_continue(x,10**(-15)) for x in values]
t1_end=time.perf_counter()
time_fractii=t1_end - t1_start


t2_start=time.perf_counter()
results_polinom=[preprocesare(x) for x in values]
t2_end=time.perf_counter()
time_polinom=t2_end-t2_start


t3_start=time.perf_counter()
results_real=[tan(x) for x in values]
t3_end=time.perf_counter()
time_library=t3_end-t3_start


errors_fractii=[abs(results_fractii[i]-results_real[i]) for i in range(n)]
errors_polinom=[abs(results_polinom[i]-results_real[i]) for i in range(n)]

print("FRACTII CONTINUE:")
print(f"Timp total: {time_fractii:.4f} secunde")
print(f"Eroare maxima: {max(errors_fractii):.2e}")
print()
print("POLINOM:")
print(f"Timp total: {time_polinom:.4f} secunde")
print(f"Eroare maxima: {max(errors_polinom):.2e}")
print()
print("BIBLIOTECA:")
print(f"Timp total: {time_library:.4f} secunde")
print()
print(f"Fractii continue vs biblioteca: {time_fractii/time_library:.2f}x")
print(f"Polinom vs biblioteca: {time_polinom/time_library:.2f}x")
```
